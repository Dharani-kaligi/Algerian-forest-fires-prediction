```python
The dataset is converted into two sets based on Region from 122th index, we can make a new column based on the Region

1 : "Bejaia Region Dataset"

2 : "Sidi-Bel Abbes Region Dataset"

Add new column with region
```

```python
df[['month','day','year','Temperature','RH','Ws']] = df[['month','day','year','Temperature','RH','Ws']].astype(int)

df.info()
```

```python
df[["Rain","FFMC","DMC","DC","ISI","BUI","FWI"]] = df[["Rain","FFMC","DMC","DC","ISI","BUI","FWI"]].astype(float)
df.info()
```

# Monthly fire analysis


```python
df['Classes'] = np.where(df['Classes'].str.contains('not fire'), 'not fire', 'fire' )
```

```python
dfmonthly =df.loc[df["Region"]==1]
plt.subplots(figsize = (13,6))
sns.set_style('whitegrid')
sns.countplot(x='month', hue= 'Classes', data = dfmonthly)
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight='bold')
plt.title("Fire Analysis of Sidi- Bel Regions",weight='bold')
```

```python
df_monthly =df.loc[df["Region"]==0]
plt.subplots(figsize = (13,6))
sns.set_style('whitegrid')
sns.countplot(x='month', hue= 'Classes', data = df_monthly)
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight='bold')
plt.title("Fire Analysis of Bjeaiah Regions",weight='bold')
```
